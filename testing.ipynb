```python
from openai import OpenAI
import data_filteration
```

```python
dataset= data_filteration.data_load()
filtered_data_title=[]
for word in dataset:
    filtered_data_title.append(word['title'])
```

```python
print(len(dataset))
print(len(filtered_data_title))
```

```python
client= OpenAI()
response = client.embeddings.create(
    input= filtered_data_title,
    model="text-embedding-3-small"
)
```

```python
response.data[0].embedding
```

```python
dataset_guid_embed=[]
dataset_embedding=[]
for i in range(0,len(filtered_data_title)):
    dataset_guid_embed.append(dataset[i]['guid'])
    dataset_embedding.append(response.data[i].embedding)
```

```python
import numpy as np
```

```python
np.savez_compressed(
    "dataset_embedding.npz",
    guids=np.array(dataset_guid_embed),
    embeddings=np.array(dataset_embedding)
)
```

```python
from openai import OpenAI
client= OpenAI()
import numpy as np
```

```python
response = client.embeddings.create(
    input=query,
    model="text-embedding-3-small"
)
```

```python
data = np.load("dataset_embedding.npz")
```

```python
data['guids'][0]
```

```python
query_embedding=response.data[0].embedding
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(
    [query_embedding],
    data['embeddings']
)[0]
```

```python
similarity = enumerate(similarity)
```

```python
similarity= sorted(similarity,key=lambda x:x[1],reverse=True)
```

```python
result=[]
for i in range(0,5):
    index,probability = similarity[i]
    print(index)
    print(probability)
    result.append(data['guids'][index])
```
